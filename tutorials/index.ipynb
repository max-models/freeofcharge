# Examples

Each notebook is executed when the documentation is built; download it from its page to run it yourself. They build on each other, from the equilibrium to an injector design scan.

1. [Equilibria and flux coordinates](/freeofcharge/examples/equilibrium-and-flux-surfaces/): the Solov'ev solution and G-EQDSK files, `rho`, the field, flux shells and the wall.
2. [Beam markers](/freeofcharge/examples/beam-markers/): aiming with `tangential_injection`, width, divergence and halo, energy components.
3. [Stopping and shine-through](/freeofcharge/examples/stopping-and-shine-through/): Suzuki's cross sections, a check against Beer-Lambert, the shine-through against density and energy, birth profiles and wall hits.
4. [Slowing down, heating and current drive](/freeofcharge/examples/slowing-down-and-current-drive/): the critical energy, the split of the heating, fast-ion pressure and the beam-driven current.
5. [Designing the injection](/freeofcharge/examples/injector-design-scan/): beam energy and tangency radius against shine-through and current-drive efficiency.
6. [Stellarator injection](/freeofcharge/examples/stellarator-injection/): a GVEC equilibrium, flux shells in 3D, deposition and the local slowing down.

The [Scripts](/freeofcharge/example-scripts/) are complete Python scripts with their plots: an ITER-like injection, the injector chain and the ASCOT5 benchmark.

See [Building the docs](/freeofcharge/development/docs/) to add an example.

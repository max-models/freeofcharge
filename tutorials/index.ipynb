# Tutorials

Each notebook is executed when the documentation is built; download it from its page to run it yourself. They build on each other, from the equilibrium to an injector design scan.

1. [Equilibria and flux coordinates](/freeofcharge/tutorials/equilibrium-and-flux-surfaces/): the Solov'ev solution and G-EQDSK files, `rho`, the field, flux shells and the wall.
2. [Beam markers](/freeofcharge/tutorials/beam-markers/): aiming with `tangential_injection`, width, divergence and halo, energy components.
3. [Stopping and shine-through](/freeofcharge/tutorials/stopping-and-shine-through/): Suzuki's cross sections, a check against Beer-Lambert, the shine-through against density and energy, birth profiles and wall hits.
4. [Slowing down, heating and current drive](/freeofcharge/tutorials/slowing-down-and-current-drive/): the critical energy, the split of the heating, fast-ion pressure and the beam-driven current.
5. [Designing the injection](/freeofcharge/tutorials/injector-design-scan/): beam energy and tangency radius against shine-through and current-drive efficiency.

See [Building the docs](/freeofcharge/development/docs/) to add a tutorial.

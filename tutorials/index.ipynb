# Examples

Each notebook is executed when the documentation is built; download it from its page to run it yourself. They build on each other, from the equilibrium to an injector design scan.

1. [Equilibria and flux coordinates](/freeofcharge/examples/equilibrium-and-flux-surfaces/): the Solov'ev solution and G-EQDSK files, $\rho$, the field, flux shells and the wall.
2. [Beam markers](/freeofcharge/examples/beam-markers/): aiming with `tangential_injection`, width, divergence and halo, energy components.
3. [Stopping and shine-through](/freeofcharge/examples/stopping-and-shine-through/): Suzuki's cross sections, a check against Beer-Lambert, the shine-through against density and energy, birth profiles and wall hits.
4. [Slowing down, heating and current drive](/freeofcharge/examples/slowing-down-and-current-drive/): the critical energy, the split of the heating, fast-ion pressure and the beam-driven current.
5. [Designing the injection](/freeofcharge/examples/injector-design-scan/): beam energy and tangency radius against shine-through and current-drive efficiency.
6. [Stellarator injection](/freeofcharge/examples/stellarator-injection/): a GVEC equilibrium, flux shells in 3D, deposition and the local slowing down.
7. [Trapping and pitch modes](/freeofcharge/examples/trapping-and-pitch-modes/): editing G-EQDSK files, trapped fractions, the pitch modes of the scattering and the trapped-electron correction (Sauter, Start-Cordey).
8. [W7-X with PolarFEEC](/freeofcharge/examples/w7x-polarfeec/): DESC's W7-X equilibrium through PolarFEEC.jl, in 3D, its inverse map against DESC, flux shells and a beam's deposition.
9. [Stellarator coils](/freeofcharge/examples/stellarator-coils/): a CoilStellaration coilset, its vacuum field and field error, and the injection lines that reach the plasma between the coils.
10. [W7-X deposition](/freeofcharge/examples/w7x-deposition/): DESC's W7-X equilibrium on a cached grid, W7-X-like injector boxes (stand-in geometry), shine-through and birth profiles per source and energy component, checked against DESC directly, and the slowing down on the 3D shells: heating, fast-ion density and pressure, the power balance and volume integrals in physical space.
11. [Fast-ion orbits](/freeofcharge/examples/fast-ion-orbits/): guiding-centre orbits in a tokamak (energy, canonical momentum, the width of a passing orbit), the orbits with Coulomb collisions against the local slowing down, and the stand-in W7-X births followed to a W7-X-like wall: prompt losses, losses during the slowing down and the wall-load map.
12. [Beam torque](/freeofcharge/examples/nbi-torque/): the collisional and J×B torque profiles, co and counter injection, and the torque per MW against beam energy.
13. [Neutron rates](/freeofcharge/examples/neutron-rates/): beam–target and thermonuclear D–D neutrons from the slowing down (Bosch–Hale reactivities), against temperature and beam energy.
14. [Fast-ion distribution](/freeofcharge/examples/fast-ion-distribution/): f(ρ, E, ξ) from the local slowing down and from the orbits, the slowing-down spectrum and the anisotropy.
15. [Beam modulation](/freeofcharge/examples/time-dependent-slowing-down/): switch-on and switch-off, a 10 ms blip and modulated power, with the response of heating, stored energy and current.
16. [Shine-through loads](/freeofcharge/examples/shine-through-wall-loads/): the unabsorbed beam on the first wall, its footprint and peak heat flux, in a tokamak and on the W7-X-like wall.
17. [Stellarator current drive](/freeofcharge/examples/stellarator-current-drive/): the effective trapped fraction of W7-X's surfaces and the beam-driven current with the trapped-electron correction.
18. [Stopping tables](/freeofcharge/examples/stopping-tables/): tabulated stopping cross sections, from a model or from ADAS adf21 files, their accuracy and speed.
19. [Charge-exchange losses](/freeofcharge/examples/charge-exchange-losses/): fast ions lost on background neutrals during the slowing down, against the edge neutral density, and the fast-neutral spectrum.
20. [Orbit width](/freeofcharge/examples/orbit-width/): the births averaged over their first orbit, prompt losses, and zero width against orbit following and ASCOT5.
21. [One-call run](/freeofcharge/examples/run-nbi/): `run_nbi` from the beams to one Dataset, the parameter file and the `freeofcharge` command.

The [Scripts](/freeofcharge/example-scripts/) are complete Python scripts with their plots: an ITER-like injection, the injector chain and the ASCOT5 benchmark.

See [Building the docs](/freeofcharge/development/docs/) to add an example.
